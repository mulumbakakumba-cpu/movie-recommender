```python
import ast

def extract_names(value):
    try:
        items = ast.literal_eval(value)
        return [item["name"] for item in items]
    except (ValueError, TypeError, SyntaxError):
        return []
```

```python
df["genres"] = df["genres"].apply(extract_names)
df["keywords"] = df["keywords"].apply(extract_names)
```

```python
display(
    df[["title", "genres", "keywords"]].head()
)
```

```python
def extract_cast(value):
    try:
        items = ast.literal_eval(value)
        return [
            item["name"]
            for item in items[:3]
        ]
    except (ValueError, TypeError, SyntaxError):
        return []
```

```python
df["cast"] = df["cast"].apply(extract_cast)
```

```python
def extract_director(value):
    try:
        items = ast.literal_eval(value)

        for item in items:
            if item.get("job") == "Director":
                return [item["name"]]

        return []

    except (ValueError, TypeError, SyntaxError):
        return []
```

```python
df["crew"] = df["crew"].apply(extract_director)
```

```python
indices = pd.Series(
    df.index,
    index=df["title"]
).drop_duplicates()
```

```python
def recommend_movies(title, n=10):
    if title not in indices:
        return f"Movie '{title}' not found."

    idx = indices[title]

    similarity_scores = list(
        enumerate(similarity_matrix[idx])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similarity_scores = similarity_scores[1:n + 1]

    movie_indices = [
        index
        for index, score in similarity_scores
    ]

    return df.iloc[movie_indices][
        ["title", "vote_average", "popularity"]
    ]
```

```python
from surprise import Dataset, Reader, SVD
```

```python
svd_ratings = ratings_tmdb[
    ["userId", "tmdbId", "rating"]
].copy()

svd_ratings = svd_ratings.rename(
    columns={"tmdbId": "movieId"}
)

print("SVD dataset shape:", svd_ratings.shape)
display(svd_ratings.head())
```

```python
reader = Reader(
    rating_scale=(0.5, 5.0)
)

surprise_data = Dataset.load_from_df(
    svd_ratings[
        ["userId", "movieId", "rating"]
    ],
    reader
)
```

```python
from surprise.model_selection import train_test_split

trainset, testset = train_test_split(
    surprise_data,
    test_size=0.2,
    random_state=42
)

print("Training ratings:", trainset.n_ratings)
print("Test ratings:", len(testset))
```

```python
svd_model = SVD(
    n_factors=100,
    n_epochs=20,
    lr_all=0.005,
    reg_all=0.02,
    random_state=42
)

svd_model.fit(trainset)

print("✅ SVD model trained")
```

```python
from surprise import accuracy

predictions = svd_model.test(testset)

rmse = accuracy.rmse(
    predictions,
    verbose=False
)

mae = accuracy.mae(
    predictions,
    verbose=False
)

print(f"RMSE: {rmse:.4f}")
print(f"MAE:  {mae:.4f}")
```

```python
def get_svd_recommendations(user_id, n=10):
    movie_ids = df["id"].tolist()

    predictions = []

    for movie_id in movie_ids:
        prediction = svd_model.predict(
            user_id,
            int(movie_id)
        )

        predictions.append({
            "movie_id": movie_id,
            "svd_score": prediction.est
        })

    recommendations = pd.DataFrame(predictions)

    recommendations = recommendations.sort_values(
        "svd_score",
        ascending=False
    ).head(n)

    recommendations = recommendations.merge(
        df[["id", "title", "vote_average", "popularity"]],
        left_on="movie_id",
        right_on="id",
        how="left"
    )

    return recommendations[
        ["movie_id", "title", "svd_score",
         "vote_average", "popularity"]
    ]
```

```python
def min_max_normalize(series):
    min_value = series.min()
    max_value = series.max()

    if max_value == min_value:
        return pd.Series(
            0.0,
            index=series.index
        )

    return (
        (series - min_value)
        / (max_value - min_value)
    )
```

```python
def hybrid_recommend_movies(
    user_id,
    title,
    n=10,
    content_weight=0.6,
    collaborative_weight=0.4
):
    if title not in indices:
        return f"Movie '{title}' not found."

    movie_idx = indices[title]

    # Content-based similarity
    content_scores = similarity_matrix[movie_idx]

    content_df = pd.DataFrame({
        "movie_id": df["id"].values,
        "content_score": content_scores
    })

    # Collaborative predictions
    collaborative_scores = []

    for movie_id in df["id"]:
        prediction = svd_model.predict(
            user_id,
            int(movie_id)
        )

        collaborative_scores.append(
            prediction.est
        )

    content_df["collaborative_score"] = (
        collaborative_scores
    )

    # Normalize scores
    content_df["content_score"] = (
        min_max_normalize(
            content_df["content_score"]
        )
    )

    content_df["collaborative_score"] = (
        min_max_normalize(
            content_df["collaborative_score"]
        )
    )

    # Hybrid score
    content_df["hybrid_score"] = (
        content_weight
        * content_df["content_score"]
        +
        collaborative_weight
        * content_df["collaborative_score"]
    )

    # Remove selected movie
    content_df = content_df[
        content_df["movie_id"] != df.iloc[movie_idx]["id"]
    ]

    # Top recommendations
    recommendations = (
        content_df
        .sort_values(
            "hybrid_score",
            ascending=False
        )
        .head(n)
    )

    recommendations = recommendations.merge(
        df[
            ["id", "title",
             "vote_average",
             "popularity"]
        ],
        left_on="movie_id",
        right_on="id",
        how="left"
    )

    return recommendations[
        [
            "title",
            "hybrid_score",
            "content_score",
            "collaborative_score",
            "vote_average",
            "popularity"
        ]
    ]
```

```python
hybrid_recommend_movies(
    user_id=1,
    title="Avatar",
    n=10
)
```
